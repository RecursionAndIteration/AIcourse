```python
import tensorflow as tf
import numpy as np
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
from tensorflow.keras.optimizers import SGD
layer1 = Dense(units=1,input_shape=[3])
layer2=Dense(units=1)
layer3=Dense(units=1)
model=Sequential([layer1,layer2,layer3])
model.compile(optimizer=SGD(learning_rate=0.001),loss="mean_squared_error")
```

```python
x=np.array([[5,9,0],[1,1,0],[6,8,0],[6,4,0],[9,7,0]],dtype=int)
y=np.array([14,2,14,10,16],dtype=int)
model.fit(x,y,epochs=400)
```

```python
print(model.predict(np.array([[10,12]])))
weights1=layer1.get_weights()
weights2=layer2.get_weights()
print(weights1)
print(weights2)
#weights=layer.get_weights()
#print(weights)
#print(f"wzor prostej to y={weights[0][0]}x+{weights[1][0]}")
```
