```python
import numpy as np
import matplotlib.pyplot as plt

# Zakres temperatury
x=np.linspace(0,50,500)

# Funkcja przynależności trójkątna: "ciepło"
def fuzzy_warm(x):
  return np.maximum(np.minimum((x-10)/20,(40-x)/10),0)
  #if x<=10 or x>=40:
    #return 0
  #elif 10<x<=30:
    #return (x-10)/(30-10)
  #else 30<x<40:
    #return (40-x)/(40-30)


# Obliczenie przynależności
jakcieplo=fuzzy_warm(x)

# Punkty charakterystyczne trójkąta
x_points=[10,30,40]
y_points=fuzzy_warm(np.array(x_points))
labels = ['a (10°C)', 'b (30°C)', 'c (40°C)']
# Rysowanie wykresu
plt.figure(figsize=(10, 5))
plt.plot(x,jakcieplo,label='cieplo',color='orange')
plt.scatter(x_points,y_points,color='black')

# Dodanie etykiet do punktów
for i, label in enumerate(labels):
  plt.annotate(label,(x_points[i],y_points[i]+0.05))

plt.title('Funkcja przynależności: Ciepło (trójkąt: 10, 30, 40)')
plt.xlabel('Temperatura [°C]')
plt.ylabel('Przynależność')
plt.grid(True)
plt.legend()
plt.ylim(0, 1.1)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
# zakres temp
x=np.linspace(0,50,500)


# Funkcja trapezowa
def fuzzy_comfort(x):
  return np.maximum(np.minimum(np.minimum((x-10)/(10),1),(40-x)/10),0)
komfort=fuzzy_comfort(x)
# Punkty charakterystyczne
x_points=np.array([10,20,30,40])
y_points=fuzzy_comfort(x_points)
labels = ['a (10°C)', 'b (20°C)', 'c (30°C)', 'd (40°C)']

# Rysowanie wykresu
plt.figure(figsize=(10, 5))
plt.plot(x,komfort,label='cieplo',color='green')
plt.scatter(x_points,y_points,color='black')


# Etykiety
for i, label in enumerate(labels):
  plt.annotate(label,(x_points[i],y_points[i]+0.05),ha='center')
plt.title('Funkcja przynależności: Temperatura komfortowa (trapez: 10, 20, 30, 40)')
plt.xlabel('Temperatura [°C]')
plt.ylabel('Przynależność')
plt.grid(True)
plt.legend()
plt.ylim(0, 1.1)
plt.show()
```
