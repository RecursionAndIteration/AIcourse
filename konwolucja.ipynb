```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def apply_filter(grayscale_image, filter_kernel):
  image_height, image_width=grayscale_image.shape

  filtered_image=np.zeros((image_height-2,image_width-2),dtype=np.float32)
  for y in range(1,image_height-1):
    for x in range(1,image_width-1):
      neighborhood=grayscale_image[y-1:y+2,x-1:x+2]
      convolution_value=np.sum(neighborhood*filter_kernel)
      filtered_image[y-1,x-1]=convolution_value # Corrected indexing here
  filtered_image=np.abs(filtered_image)
  filtered_image=np.clip(filtered_image,0,255).astype(np.uint8)
  plt.figure(figsize=(10,4))
  plt.subplot(1,2,1)
  plt.title("oryginalny obraz")
  plt.imshow(grayscale_image,cmap='gray')
  plt.subplot(1,2,2)
  plt.title("obraz po filtracji")
  plt.imshow(filtered_image,cmap='gray')
  plt.show()
```

```python
apply_filter(img_grey,np.array([
    [-1,0,1],
    [-2,0,2],
    [-1,0,1]
]))
#krawedzie pionowe
```

```python
apply_filter(img_grey,np.array([
    [-1,-2,-1],
    [0,0,0],
    [1,2,1]
]))
#krawedzie poziome
```

```python
#krawedzie ogolne
apply_filter(img_grey,np.array([
    [0,-1,0],
    [-1,4,-1],
    [0,-1,0]
]))
```

```python
#wyostrzenie
apply_filter(img_grey,np.array([
    [0,-1,0],
    [-1,5,-1],
    [0,-1,0]
]))
```

```python
#filtr filtr filtr
apply_filter(img_grey,np.array([
    [1,0,0],
    [0,1,0],
    [0,0,1]
]))
```
