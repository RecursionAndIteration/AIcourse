```python
import json,random,math
def get_words_from_text(text):


  punctuation = r'.,!?;:"()[]{}<>-/\_@#$%^&*+=`~\'–—‚„‟'


  clean_words = []
  current_word = []
  for char in text:
    if char.isalnum() or char in 'ąćęłńóśźżĄĆĘŁŃÓŚŹŻ': # Sprawdzamy czy znak jest alfanumeryczny lub polską literą diakrytyczną
      current_word.append(char)
    else:
      if current_word:
        clean_words.append(''.join(current_word).lower())
        current_word = []

  if current_word: # Dodaj ostatnie słowo, jeśli istnieje
    clean_words.append(''.join(current_word).lower())

  return clean_words

def wczytajdane(path1,path2):
  f1 = open(path1, encoding="utf-8")
  f2 = open(path2, encoding="utf-8")
  data = json.load(f1)+json.load(f2)
  for mail in data:
    mail["text"]=get_words_from_text(mail["text"])
  return data

def train_test_split(data, test_ratio = 0.2):
  random.seed(0)
  random.shuffle(data)
  cut = int(len(data)* test_ratio)
  print(f"Ilość dla zbioru uczącego: {len(data[cut:])}")
  print(f"Ilość dla zbioru testowego: {len(data[:cut])}")
  return data[cut:], data[:cut]


def build_vocab(train):
  vocab = set()
  for rec in train:
    vocab.update(rec["text"])
  print(vocab)
  return vocab

def train_nb(train, vocab, alpha=1.0):
  class_counts = {}
  word_counts = {}
  total_words = {}
  for rec in train:
    label = rec["label"]
    class_counts.setdefault(label, 0)
    word_counts.setdefault(label, {})
    total_words.setdefault(label, 0)
    for word in rec["text"]:
      word_counts[label].setdefault(word,0)
  for rec in train:
    label = rec["label"]
    class_counts[label]+=1

    for word in rec["text"]:
      word_counts[label][word]+=1
      total_words[label]+=1
  model = {
      "class_counts": class_counts,
      "word_counts": word_counts,
      "total_words": total_words,
      "vocab": vocab,
      "alpha": alpha,
      "total_docs": len(train)
  }
  print(model)
  return model

def log_prob(model, rec, class_name):
  logp = math.log(model["class_counts"][class_name]/model["total_docs"])
  V = len(model["vocab"])
  a = model["alpha"]


  for tag in rec["text"]:
    wc = model["word_counts"][class_name].get(tag,0)
    logp += math.log((wc+a)/(model["total_words"][class_name]+a*V))
  return logp

def predict(model,rec):
  best_class, best_log = None, -1e99
  for label in model["class_counts"]:
    lp = log_prob(model, rec, label)
    if lp > best_log:
      best_log = lp
      best_class = label
  return best_class

def evaluate(model,test):
  correct = 0
  for rec in test:
    if predict(model,rec) == rec["label"]:
      correct+=1
  accuracy = correct/len(test)
  print(f"Dokładność: {accuracy:.2%}")
```

```python
def main():
  path1="ham_emails.json"
  path2="spam_emails.json"
  data = wczytajdane("ham_emails.json","spam_emails.json")
  train,test = train_test_split(data)
  vocab = build_vocab(train)
  model=train_nb(train,vocab,alpha=1.0)
  sample=test[0]
  print("\n przykladowe tagi : ", sample["text"])
  print("\n rzeczywiste:",sample["label"])
  print("\n przewidziane:",predict(model,sample))
  evaluate(model,test)
main()
```
